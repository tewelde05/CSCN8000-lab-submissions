##Stacking and Splitting

```python
arr = np.array([[1,3],[6,7]])
print("arr:\n", arr)
arr1 = np.array([[1,2],[3,4]])
print("arr1:\n", arr1)
vertical_stack = np.vstack((arr, arr1))
print("vertical_stack:\n", vertical_stack)
horizontal_stack = np.hstack((arr, arr1))
print("horizontal_stack:\n", horizontal_stack)
split_arr = np.hsplit(horizontal_stack, 2)
print("split_arr:\n", split_arr)
split_arr1 = np.vsplit(vertical_stack, 2)
print("split_arr1:\n", split_arr1)
```

## Broadcasting

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

values = np.array([1, 2, 3])

subtraction = matrix - values
multiplication = matrix * values

print("Original matrix:")
print(matrix)

print("\nSubtraction using broadcasting:")
print(subtraction)

print("\nMultiplication using broadcasting:")
print(multiplication)
```

## Systems of Linear Equations

```python
A1 = np.array([
    [2, 3, -4],
    [1, -4, 0]
], dtype=float)

b1 = np.array([6, 8], dtype=float)

rank_A1 = np.linalg.matrix_rank(A1)
rank_aug1 = np.linalg.matrix_rank(np.column_stack((A1, b1)))
variables1 = A1.shape[1]

print("Rank of A:", rank_A1)
print("Rank of augmented matrix:", rank_aug1)
print("Number of variables:", variables1)

if rank_A1 != rank_aug1:
    print("No solution")
elif rank_A1 == variables1:
    print("One unique solution")
else:
    print("Infinitely many solutions")

solution1 = np.linalg.lstsq(A1, b1, rcond=None)[0]
print("One solution returned by NumPy:", solution1)
print("Because there are 3 unknowns but only 2 independent equations, the system has infinitely many solutions.")
```
